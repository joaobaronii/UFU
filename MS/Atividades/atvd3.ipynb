5. Uma fábrica possui quatro máquinas idênticas e críticas. O tempo de vida de cada máquina é exponencial com média de cem dias. A fábrica emprega um técnico para consertá-las. O tempo de reparo é exponencial com média de dois dias. Modele o número de máquinas funcionando como uma CMTC.

    (b) Qual a probabilidade a longo prazo de todas as máquinas estarem quebradas?

    (c) Qual o número médio de máquinas operantes?  

```python
import numpy as np

def simular_fabrica(dias_simulacao=100000):
    estado = 4 
    tempo_atual = 0.0
    tempo_nos_estados = np.zeros(5)
    
    while tempo_atual < dias_simulacao:
        taxa_falha = estado * 0.01
        taxa_reparo = 0.5 if estado < 4 else 0.0
        
        taxa_total = taxa_falha + taxa_reparo
        if taxa_total == 0:
            break   
    
        dt = np.random.exponential(1.0 / taxa_total)
    
        if tempo_atual + dt > dias_simulacao:
            tempo_nos_estados[estado] += (dias_simulacao - tempo_atual)
            break
            
        tempo_nos_estados[estado] += dt
        tempo_atual += dt
        
        prob_falha = taxa_falha / taxa_total
        if np.random.rand() < prob_falha:
            estado -= 1
        else:
            estado += 1
            
    probabilidades = tempo_nos_estados / dias_simulacao
    
    maquinas_esperadas = np.sum(np.arange(5) * probabilidades)
    
    return probabilidades, maquinas_esperadas

probs_5, media_5 = simular_fabrica()

print("(b) Probabilidade de todas quebradas:", probs_5[0])
print("(c) Número médio de máquinas operantes:", media_5)
```

8. Um sistema é formado por dois componentes idênticos. Um único reparador. Cada componente funciona por um tempo exponencial com taxa λ = 1/mês e é reparado em um tempo exponencial com taxa μ = 3/mês. Modele o número de componentes funcionais X(t) como uma CMTC com estados {0, 1, 2}. Escreva a matriz G e encontre a distribuição limitante.

```python
def simular_sistema_8(meses_simulacao=100000):
    np.random.seed(42)
    estado = 2
    tempo_atual = 0.0
    tempo_nos_estados = np.zeros(3) 
    
    while tempo_atual < meses_simulacao:
        taxa_falha = estado * 1.0
        taxa_reparo = 3.0 if estado < 2 else 0.0
        
        taxa_total = taxa_falha + taxa_reparo
        if taxa_total == 0:
            break
            
        dt = np.random.exponential(1.0 / taxa_total)
        
        if tempo_atual + dt > meses_simulacao:
            tempo_nos_estados[estado] += (meses_simulacao - tempo_atual)
            break
            
        tempo_nos_estados[estado] += dt
        tempo_atual += dt
        
        prob_falha = taxa_falha / taxa_total
        if np.random.rand() < prob_falha:
            estado -= 1 
        else:
            estado += 1 
            
    probabilidades = tempo_nos_estados / meses_simulacao
    return probabilidades

probs_8 = simular_sistema_8()

print("Distribuição Limitante:")
print(f"P(X=0) = {probs_8[0]:.4f}")
print(f"P(X=1) = {probs_8[1]:.4f}")
print(f"P(X=2) = {probs_8[2]:.4f}")
```
